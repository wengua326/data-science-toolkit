```python
#1. Create a numpy array of size 2 x 2
import numpy as np
array22=np.array([[2,4,],[5,7]])
print("This is array size 2*2")
print(array22)
```

```python
#Create a random numpy array of size 3 x 4
array34=np.random.rand(3,4)
print("This is array size 3*4")
print(array34)
```

```python
#Create a numpy array of ones and a numpy array of zeroes
one=np.ones((3,4))
zero=np.zeros((3,4))
print("This is array of ones")
print(one)
print("This is array of zeros")
print(zero)
```

```python
#Create a random numpy array of size 6x3 and transpose it
tarray=np.random.rand(6,3)
print("This is random array")
print(tarray)
print("This is transpose matrix")
print(np.transpose(tarray))
```

```python
#Create two random numpy array of size n x p and p x m and then compute the dot product
random1=np.random.rand(2,3)
random2=np.random.rand(3,4)
print(random1)
print(random2)
print("this is dot product")
print(np.dot(random1,random2))
```
