Question 1: Identify the Problem
Consider this dataset:
Student Math Science English
Ali 75 80 72
Sara 88 91 85
John 65 70 68
1. Is this data in tidy format?
2. If not, explain which tidy-data rule is violated.
3. Using Python/pandas, reshape the dataset so that it has the columns:
- Student
- Subject
- Score


1.   No, this dataset is NOT in tidy format.
2.   Each variable forms a column




```python
import pandas as pd
data_q1 = {
    "Student": ["Ali", "Sara", "John"],
    "Math": [75, 88, 65],
    "Science": [80, 91, 70],
    "English": [72, 85, 68],
}
df_q1 = pd.DataFrame(data_q1)
print(df_q1)
print("")
df_melted = pd.melt(df_q1, id_vars=["Student"], var_name="Subject", value_name="Score")
print(df_melted)
```

Question 2: Sales Data
You are given:
Country Q1_Sales Q2_Sales Q3_Sales Q4_Sales
Malaysia 12000 15000 14000 18000
Thailand 10000 11000 13000 16000
Indonesia 14000 12500 15000 17000
1. What is the tidy-data problem with this dataset?
2. Use pandas to convert it into:
Country Quarter Sales
Malaysia Q1 12000
Malaysia Q2 15000

1.Not all each variable form a column

```python
import pandas as pd
data_q2 = {
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000],
}
df_q2 = pd.DataFrame(data_q2)
print(df_q2)
print("")
df_q2_tidy = pd.melt(
    df_q2,
    id_vars= ["Country"],
    var_name="Sales",
    value_name="Result"
)
print(df_q2_tidy)
```

Question 3: Asking Questions from Data
Consider this dataset:
data = { "Student": ["A", "B", "C", "D", "E", "F"], "Study_Hours": [2, 5, 3, 7, 4, 8],
"Score": [55, 72, 60, 85, 68, 90]}
Using Python:
1. What is the average study time?
2. What is the average score?
3. Which student obtained the highest score?
4. Which student studied the most?
5. Question: Does studying more hours appear to be associated with a higher score?
6. Create a scatter plot of Study_Hours vs Score.


```python
data = { "Student": ["A", "B", "C", "D", "E", "F"],
      "Study_Hours": [2, 5, 3, 7, 4, 8],
      "Score": [55, 72, 60, 85, 68, 90]
     }
df_q3 = pd.DataFrame(data)
print(df_q3)
print("Average study time are ", df_q3["Study_Hours"].mean().round(2),"hours")
print("Average scores are ", df_q3["Score"].mean().round(2))
print("The student that obtain highest score is ",df_q3.loc[df_q3["Score"].idxmax(),"Student"])
print("The student that study the most is ",df_q3.loc[df_q3["Study_Hours"].idxmax(),"Student"])
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(df_q3["Study_Hours"],df_q3["Score"],color="red",s=10,alpha=0.9)
for i in range(len(df_q3)):
  plt.text(
      df_q3["Study_Hours"][i]+0.12,
      df_q3["Score"][i]-0.5,
      df_q3["Student"][i],
      fontsize=15
  )

plt.title("Score vs Study Hour")
plt.xlabel("Study Hour")
plt.ylabel("Score")
plt.grid(True,alpha=0.6)
plt.show()
```
