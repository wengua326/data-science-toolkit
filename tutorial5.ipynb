```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/salary.csv')
display(df.head())
```

```python
print(df.columns[6])
df.rename(columns={df.columns[6]: 'Salary'},inplace=True)
display(df.head())
```

dtype : object

```python
df["Salary"] = df["Salary"].replace(",","",regex=True)
df["Salary"] = df["Salary"].astype(int)
df.head()
```

```python
df.info()
```

```python
df.rename(columns={df.columns[7]: 'Compensation'},inplace=True)
display(df.head())
print("Sum of null:",df["Compensation"].isnull().sum())
```

```python
#too many missing values and too many nans, not useful then drop
df.rename(columns={df.columns[5]: 'Context'},inplace=True)
df.rename(columns={df.columns[9]: 'Other currency'},inplace=True)
df.rename(columns={df.columns[10]: 'Salary context'},inplace=True)
df = df.drop(
    columns=[
        "Other currency",
        "Salary context",
        "Context"
    ])
df.head()

```

```python
df_clean = df.dropna()
#if you want to see dropped row
#~this symbol will find the indexs not in df_clean
dropped_rows = df[~df.index.isin(df_clean.index)]
print("Deleted row:")
print(dropped_rows)
```

```python
#onlty drop rows containing at least 2 missing value
df.dropna(thresh=2)
df.dropna(how='all')
row_before = len(df)
df_clean = df.dropna()
row_after = len(df_clean)
print("Row before: ",row_before)
print("Row after: ",row_after)
print("Deleted rows:",row_before-row_after)
```

imputation(missing data)

```python
#import the library
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="constant",fill_value=0)
df_imputed = df.copy()

for col in df_imputed.columns:
  if df_imputed[col].isnull().any():
    df_imputed[col] = imputer.fit_transform(df_imputed[[col]]).ravel()

#so your imputed column is now filled with the most common value
print(df_imputed.isna().sum())
df_imputed.head()
```

duplicated value

```python
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

```python
df_imputed.drop(columns=[
  'Unnamed: 0',
  'Timestamp'
],inplace=True)
df_imputed.rename(columns={df_imputed.columns[6]:"Country"},inplace=True)
df_imputed.rename(columns={df_imputed.columns[7]:"Other_Country"},inplace=True)
df_imputed.head()
```

```python
from rapidfuzz import process
display(df_imputed["Country"])
x=df_imputed['Country'].astype(str).str.isnumeric()
df_imputed.loc[x,'Country']="USA"
standard = ["USA","UK"]

def scan (country):
  match,score,_ = process.extractOne(str(country),standard)
  return match if score>70 else country

df_imputed['Country']=df_imputed['Country'].apply(scan)
display(df_imputed["Country"])
```

noisy data

```python
df_imputed["Bin"]=pd.qcut(df_imputed["Salary"],q=3,labels=["low","medium","high"])
display(df_imputed.head())
df_imputed["Smooth_Mean"] = df_imputed.groupby("Bin")["Salary"].transform("mean").round(2)
display(df_imputed["Smooth_Mean"])


```
